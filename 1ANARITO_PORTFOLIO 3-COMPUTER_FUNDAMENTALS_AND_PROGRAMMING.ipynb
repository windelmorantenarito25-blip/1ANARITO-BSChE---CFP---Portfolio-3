This program acts as a digital planner for organizing a traditional Camalig Pinangat feast among friends. It calculates the total budget for the ingredients and mathematically determines the maximum spice level the group can handle. By analyzing the guests' preferences, the system makes sure that they receive a perfectly scaled, authentic Camaligueño dish.

```python
# 1. THE WELCOME MENU
# (This is a void function.)
def show_menu():
    # (Using the Built-In function print() to display the welcome banner.) 
    print("=========================================")
    print("       Camalig Pinangat Feast Maker      ")
    print("=========================================")
    print("Spice Levels: 1=Mild, 2=Medium, 3=Volcanic")
    print("-----------------------------------------")

# 2. THE RECIPE COST CALCULATOR
# (This function calculates the cost. 'bundles' and 'spice' are the PARAMETERS used inside the function to catch the input.)
def calc_budget(bundles, spice):
    base_price = 45.0  # Base cost of taro leaves and coconut milk
    # (Using a multi-way decision to adjust the price of chili based on the spice parameter.)
    if spice >= 3:
        sili_cost = 15.0
    elif spice == 2:
        sili_cost = 10.0
    else:
        sili_cost = 5.0
    
    # (Calculating the total using mathematical operators.)
    base_total = base_price * bundles
    spice_total = sili_cost * bundles
    total = base_total + spice_total

    print("\n--- Math Breakdown ---")
    print("Base cost calculation:", bundles, "bundles * PHP", base_price, "= PHP", base_total)
    print("Spice cost calculation:", bundles, "bundles * PHP", sili_cost, "= PHP", spice_total)
    print("Grand total calculation:", base_total, "+", spice_total, "= PHP", total)
    
    return total
    # (The 'return' keyword makes this a fruitful function.)

## MAIN PROGRAM EXECUTION
show_menu()
# (Calling the void function to print the header.)

# 3. GATHERING BUNDLE ORDERS
str_bundles = input("How many bundles of Pinangat are being made? ")
try:
    # (String Conversion: Using the built-in int() function to convert the text string into a whole number.) 
    user_bundles = int(str_bundles)
except:
    user_bundles = 1

# 5. FINDING THE BRAVEST SPICE LEVEL
# (Gathering spice tolerances from 3 different friends to find out how spicy the batch should be.)
print("\nAsk 3 friends for their preferred spice level (1 to 3):")
f1 = int(input("Friend 1's level: "))
f2 = int(input("Friend 2's level: "))
f3 = int(input("Friend 3's level: "))
highest_spice = max(f1, f2, f3)

# (Using the built-in max() function. It looks at all the inputs and finds the highest number. This ensures the Pinangat is spicy enough for the bravest friend.) 
print("--> The highest requested spice level is:", highest_spice)

# 6. EXECUTING THE BUDGET CALCULATION 
# (Calling the fruitful function.) 
final_cost = calc_budget(user_bundles, highest_spice)

# 7. DISPLAYING THE GRAND TOTAL 
display_cost = float(final_cost)

print("\n-----------------------------------------")
print("Total estimated budget: PHP", display_cost)
print("-----------------------------------------")
```
